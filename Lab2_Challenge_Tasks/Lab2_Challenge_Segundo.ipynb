```python
target_name = "SEGUNDO"
SEED_NUM = 2
redacted_name = ""

for char in target_name:
    if char in "AEIOU":
     redacted_name += str(SEED_NUM)
    else:
     redacted_name += char

print("Original Surname:", target_name)
print("SEED_NUM Used:", SEED_NUM)
print("Redacted Output:", redacted_name)
```

```python
engineer_name = "SEGUNDO"
SEED_NUM = 2
loops = 5

ASCII_VAL = ord(engineer_name[0])
verification_hash = (ASCII_VAL * SEED_NUM) - loops

print("first Letter ASCII (A):", ASCII_VAL)
print("Verification Hash:", verification_hash)
```

```python
correct_password = "ADMIN"
attempts = 0
while True:
    pwd = input("Enter Password: ")
    if pwd == correct_password:
        print("Access Granted.")
        break
    else:
        attempts += 1
        print("Access Denied.")

print(f"Attempts Before Success: {attempts}")
```
